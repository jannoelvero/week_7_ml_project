```python
matching_tracks = track_only.merge(
    df_spotify,
    on=["track_name", "artist_name"],
    how="inner",
    suffixes=("_track", "_spotify")
)

matched_ids = matching_tracks["track_id_track"].nunique()

print("Excluded track IDs:", len(track_only))
print("Matched by track name and artist:", matched_ids)
print("No matching track and artist:", len(track_only) - matched_ids)
```

```python
matched_track_ids = matching_tracks["track_id_track"].unique()

unmatched_tracks = track_only[
    ~track_only["track_id"].isin(matched_track_ids)
]

print("Unmatched tracks:", len(unmatched_tracks))
```
