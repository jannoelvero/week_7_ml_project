### Duplicate Track Handling

Duplicate track observations were identified during data cleaning. These records are intentionally retained because the same track may appear more than once in the dataset due to different album, single, compilation, or release contexts.

Removing these observations at this stage could reduce potentially useful information and would also change the distribution of the dataset, including the popularity threshold used later in the analysis.

However, retaining duplicate tracks creates a potential data leakage issue during machine learning. If a standard random train test split is used, different observations of the same track could appear in both the training and test sets. This could result in an overly optimistic estimate of model performance because the model would be evaluated on tracks that are also represented in the training data.

To prevent this, the machine learning stage will use a grouped train test split. Observations belonging to the same track will be assigned to the same group so that all versions of a track appear exclusively in either the training set or the test set.

**Decision:** Duplicate track observations are retained in the cleaned dataset and will be controlled through grouped splitting during machine learning.


### Investigation of Repeated Tracks

The initial duplicate check found no fully duplicated rows and no duplicated Spotify `track_id` values. However, the same recording may still appear under different Spotify track IDs when it is released as a single, album track, compilation track, or re release.

To investigate this possibility, repeated combinations of track name and artist name are examined separately. These observations are not automatically removed because they may represent legitimate release variations. The results will be used to determine whether grouped splitting is required during machine learning.


```python
repeated_tracks = df[
    df.duplicated(
        subset=["track_name", "artist_name"],
        keep=False
    )
].sort_values(["track_name", "artist_name"])

print("Rows with repeated track and artist combinations:", len(repeated_tracks))
print(
    "Unique repeated track and artist combinations:",
    repeated_tracks[["track_name", "artist_name"]]
    .drop_duplicates()
    .shape[0]
)
```

### Duplicate Track Decision

The dataset contains no duplicated `track_id` values and no fully duplicated rows. However, further investigation identified 1,548 observations belonging to 697 repeated track and artist combinations.

These observations represent cases where the same track appears under different Spotify track IDs, such as album, single, deluxe edition, compilation, or other release versions. Track popularity may also differ between these versions.

These observations will not be removed because they contain legitimate release level information that may be useful for the analysis. Instead, a track grouping identifier will be created using the track name and artist name.

During machine learning, a grouped train test split will be used so that all observations belonging to the same track and artist combination are assigned entirely to either the training set or the test set. This prevents different releases of the same song from appearing in both datasets and reduces the risk of data leakage.

**Decision:** Retain repeated track observations and control them through grouped train test splitting during machine learning.


```python
df["track_group"] = (
    df["track_name"].str.lower().str.strip()
    + " | "
    + df["artist_name"].str.lower().str.strip()
)
```
