The objective of this analysis is to identify the factors associated with track popularity and translate the findings into practical insights for music release and content strategy.

# Predictive Modelling Component

This project examines track popularity through two complementary approaches: an **inferential analysis**, which tests individual relationships between track characteristics and popularity, and a **predictive model**, which estimates popularity from those characteristics on data it has not seen.

The research questions above define the inferential approach, evaluated through *correlation and hypothesis testing*

This section defines the predictive approach: *a target variable, a set of predictor features, and a measure of how accurately popularity can be predicted from them*.

The two approaches answer the same questions at different levels of evidence.
-> The inferential analysis tests each relationship in isolation. 
-> The predictive model evaluates all features jointly, with each one controlled for the others so a relationship that holds in isolation may weaken or disappear once the remaining characteristics are accounted for.

**Target variable:** 'track_popularity' (0–100)

### Artist-level features

'artist_popularity' and 'artist_followers' are proposed for exclusion from the model, for two reasons:

1. **Spotify calculates artist popularity from the popularity of that artist's own tracks.** In other words, a model using artist popularity to predict track popularity would be circular, because the feature and the target come from the same source.

2. **When deciding whether to release a track by a new artist, these numbers are not available.** A model that depends on them cannot inform the decision it was built for.

This is not settled. In the cleaned dataset the correlation between 'artist_popularity' and 'track_popularity' is 0.454, which is moderate. Excluding them removes the strongest predictor in the data. The alternative is to build the model both ways and then compare.

## Mapping of the Original Research Questions

No research question is lost here. The reframing changes the way each one is answered, it does not remove any of them.

**RQ1 and RQ2** — artist popularity and follower count —> answered in the exploratory analysis, by plotting each one against track popularity and measuring the correlation between them. Those results are what justify leaving artist-level features out of the model.

**RQ3 to RQ8** — explicit content, duration, album type, album size, genre and release period -> become the model's features. Each is answered twice: individually in the exploratory analysis, and jointly through the model, where every feature is evaluated with the others controlled for.

**RQ9 and RQ10** — top artists and top tracks —> remain descriptive rankings in the exploratory analysis. They identify what is already in the data rather than predicting anything, so they fall outside the modelling scope.
